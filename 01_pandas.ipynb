```python
import pandas as pd
```

```python
#series - 1d labeled Array

s = pd.Series([1,2,3,4,5,6])
print(s)
print(type(s))
```

```python
#index - is like position ( that fixed ( like on 0th position 1 is stored))

#label - is an named index ( address of an index ) 
# EX:
s = pd.Series([54 , 65 , 85 , 96 ], index=["adam" , "Raj" , "Amit" , "kushal"]) #index is an label that signed the adam = 54
print(s)
print("---" * 5)
print(s["adam"])
print(s["kushal"])
```

```python

#Series Properties
"""
1 . Homogenous 
2 . Vectorized Operations
3 . Handle missing Values with NaN
4 . Mutable values, immutable size

"""


s1 = pd.Series([ 1 , 2 , 3 , 4 , 5])
s2 = pd.Series([10 ,20, 30 , 40 ,50])

print(s1 + s2)   #-> 1 + 10 = 11 , 2 +  20 22 ......
```

```python


# Dataframe - 2d lablled array
# -----------------Creating Dataframe using Dictinary of list 
info = {
    "name" : ["Rahul" , "Raj" , "Amit" , "keshav"],
    "Age"  : [ 21 , 22 ,24 , 19  ],
    "Cgpa" : [ 9.2 , 9.3 , 8.6 , 5.6],
}

df = pd.DataFrame(info)

print(info)
print("-----"*15)
print(df , type(df))

print(df.index)
print(df.columns)
```

```python
# --------------------------Creating DataFrames using List of List

df =pd.DataFrame([["Adam", 21] , ["Eve" , 22 ], [ "Bob" , 21]], columns = ["name" , "age"]) 
print(df)
```
