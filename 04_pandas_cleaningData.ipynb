```python


#cleaning our data
"""
1 . Handling missing values 
     - isnull() / isna()
     - isnull().sum()
     - dropna()
     - fillna(value)
     - ffill()
     - bfill()
"""
```

```python
import pandas as pd

df = pd.read_csv("raw_data.csv")


df.isnull()    # missing value - returns True
df.isna()    # missing value - returns True


df.isnull().sum()  #count the missing values in column


df.dropna()    #Drop that missing value row from table


df.fillna(0)   #stored all empty value as 0 ( not zero actually that you pass in paramter )


df.ffill()   # forward fill  -> stored a preivous value in empty space ( like 1 , , 2 , 3)
                                            #output -  ( 1 , 1 , 2 , 3 )     ------>

df.bfill()   # backward fill -> stored a last value in empty space ( like 1 , , 2 , 3)
                                            #output -  ( 1 , 2, 2 , 3 )       <-------


# df
```

```python
# Handle Duplicted Values

"""
1. dupilcated()
2. drop_dupilcates()
"""


df.duplicated()  #check Deuplicated row ---> return True

df["country"].duplicated()  #checks for speicfic rows   ---> return True

df.drop_duplicates()         # Deleted Duplicated rows
df.drop_duplicates(inplace=True)         # changes in original data 
```
