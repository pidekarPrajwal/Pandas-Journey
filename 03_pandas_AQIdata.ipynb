```python
# !pip install pandas
# AQI DataSet
import pandas as pd


df = pd.read_csv("globalAirQuality.csv")
df
```

```python
#Selecting the columns

df["city"]
df[["city" , "aqi"]]
```

```python
#selecting the rows (using label and index  - { loc and iloc })


df.loc[0]          # index 0 row data 
df.loc[0:2]       # Slicing -- StrIdx : EndIdx   ( ending index is inclusive ( means include ( if you pass 3 then from start to 3rd row will be print)))

df.iloc[2]
df.iloc[0:2]       # Slicing -- StrIdx : EndIdx   ( ending index is exclusive ( means not include ( if you pass 3 then from start to 2nd row will be print not the 3rd row)))
```

```python


#Cells  -- ( row , column)

    # row  [    "columns"    ]  
df.loc[0 , ["city", "aqi"]]   # columns in pass in [sqaure brackets]
df.loc[0:2 , ["city", "aqi"]]   # columns in pass in [sqaure brackets]


    #index --- the column value will be in integer (not in string ..) becuase i in iloc is represents index
df.iloc[0 , 2]
df.iloc[0:3 , 2]


# printing Country, city , lat and log using (loc)
df.loc[0 , ["country" , "city" , "latitude" , "longitude"]]
df.loc[0:2 , ["country" , "city" , "latitude" , "longitude"]]



# printing Country, city , lat and log using (iloc)
df.iloc[0 , [ 1, 2 , 3 , 4]]
df.iloc[0:3 , 1:5]  #using Slice
```

```python


#Selected single Scalar values (using  -- at & iat)
    
df.at[0 , "city"]    #0th row and column city

df.iat[0,2]          # 0th row and city (index == 2)
```

```python



df["city"]          # is created a view not an copy
cities=df["city"]
cities[0]  = "mumbai"    # that changes occured in original file 
 #  if you do changes in views  it will also occured in the original file 

# for that reason use .copy()   
    # it created the copy not doing changes in original file
df["city"].copy()
```

```python


#Filtering our data


df[ df ["aqi"] > 100]
df[ (df ["aqi"] > 100)  & (df["temperature"] > 30) ]   #filtering using the operators


aqi_data = [df[ (df ["aqi"] > 100)  & (df["temperature"] > 30) ] [["city" , "aqi"]]]
aqi_data
```

```python



# using Query method

df[ df ["aqi"] > 100] # instead of doing this we use query method 


# like
df.query("aqi > 100")


#rules 
    # - Expression as string ( pass the condition in string )
    # - column based Referance  ( directly wrote column name in condition)
    # - use Backticks for col_name with space or special chars
    # - Operators ( &, | , ~, >= , <= , == ,etc...)
    # - @ to referance Python vars  ( for using any value as variable in conditions)
    # - Chained Comparisons ( multiple condition in one )


df.query("aqi > 100  & temperature > 30") [["city" , "aqi"]]

# using @ variable in condition 

aqi_value = 100 
df.query("aqi > @aqi_value  & temperature > 30") [["city" , "aqi"]]

```
